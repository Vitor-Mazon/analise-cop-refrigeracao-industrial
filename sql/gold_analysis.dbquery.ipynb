```sql
-- Estatísticas por regime operacional
SELECT
  regime_operacao,
  COUNT(*) AS total_registros,
  ROUND(AVG(cop), 3) AS cop_medio,
  ROUND(AVG(vazao_m3_h), 2) AS vazao_media,
  ROUND(AVG(dT_c), 2) AS dt_medio,
  ROUND(AVG(potencia_kw), 2) AS potencia_media
FROM analytics.digital_twin.dados_refrigeracao_gold
GROUP BY regime_operacao
ORDER BY cop_medio DESC;

-- Períodos com menor eficiência
SELECT
  timestamp,
  cop,
  vazao_m3_h,
  dT_c,
  potencia_kw,
  regime_operacao,
  status_alerta,
  evento_anomalo
FROM analytics.digital_twin.dados_refrigeracao_gold
WHERE status_alerta = TRUE
ORDER BY cop ASC
LIMIT 20;

-- Monitoramento diário de eficiência
SELECT
  DATE(timestamp) AS data,
  ROUND(AVG(cop), 2) AS cop_medio,
  SUM(CASE WHEN status_alerta THEN 1 ELSE 0 END) AS total_alertas
FROM analytics.digital_twin.dados_refrigeracao_gold
GROUP BY DATE(timestamp)
ORDER BY data;
```
