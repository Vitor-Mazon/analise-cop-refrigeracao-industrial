```sql
-- Contagem por regime operacional
SELECT
  regime_operacao,
  COUNT(*) AS total_registros
FROM analytics.digital_twin.dados_refrigeracao_gold
GROUP BY regime_operacao
ORDER BY total_registros DESC;

-- Percentual de alertas operacionais
SELECT
  COUNT(*) AS total_registros,
  SUM(CASE WHEN status_alerta THEN 1 ELSE 0 END) AS total_alertas,
  ROUND(100.0 * SUM(CASE WHEN status_alerta THEN 1 ELSE 0 END) / COUNT(*), 2) AS pct_alertas
FROM analytics.digital_twin.dados_refrigeracao_gold;

-- Percentual de eventos anômalos
SELECT
  COUNT(*) AS total_registros,
  SUM(CASE WHEN evento_anomalo THEN 1 ELSE 0 END) AS total_eventos,
  ROUND(100.0 * SUM(CASE WHEN evento_anomalo THEN 1 ELSE 0 END) / COUNT(*), 2) AS pct_eventos
FROM analytics.digital_twin.dados_refrigeracao_gold;
```
